##Devolución de cambio por técnica voraz

```python
def cambio_monedas(N, SM):
  SOLUCION = [0]*len(SM)   #SOLUCION = [0,0,0,0,..]
  ValorAcumulado = 0

  for i,valor in enumerate(SM):
    monedas =  (N-ValorAcumulado)//valor
    SOLUCION[i] = monedas
    ValorAcumulado = ValorAcumulado + monedas*valor

    if ValorAcumulado == N:
      return SOLUCION


cambio_monedas(15,[25,10,5,1])
```

## N-Reinas por técnica de vuelta atrás

```python
def escribe(S):
  n = len(S)
  for x in range(n):
    print("")
    for i in range(n):
      if S[i] == x+1:
        print(" X " , end="")
      else:
        print(" - ", end="")


def es_prometedora(SOLUCION,etapa):
  #print(SOLUCION)
  #Si la solución tiene dos valores iguales no es valida => Dos reinas en la misma fila
  for i in range(etapa+1):
    #print("El valor " + str(SOLUCION[i]) + " está " +  str(SOLUCION.count(SOLUCION[i])) + " veces")
    if SOLUCION.count(SOLUCION[i]) > 1:
      return False

    #Verifica las diagonales
    for j in range(i+1, etapa +1 ):
      #print("Comprobando diagonal de " + str(i) + " y " + str(j))
      if abs(i-j) == abs(SOLUCION[i]-SOLUCION[j]) : return False
  return True



def reinas(N, solucion=[], etapa=0):
  if len(solucion) == 0:
      solucion=[0 for i in range(N)]

  for i in range(1, N+1):
    solucion[etapa] = i

    if es_prometedora(solucion, etapa):
      if etapa == N-1:
        print(solucion)
        #escribe(solucion)
        print()
      else:
        reinas(N, solucion, etapa+1)
    else:
      None

    solucion[etapa] = 0

reinas(8)
```

## Viaje por el rio. Programación dinámica

```python
TARIFAS = [
[0,5,4,3,999,999,999],
[999,0,999,2,3,999,11],
[999,999, 0,1,999,4,10],
[999,999,999, 0,5,6,9],
[999,999, 999,999,0,999,4],
[999,999, 999,999,999,0,3],
[999,999,999,999,999,999,0]
]



################################################################
def Precios(TARIFAS):
################################################################
  #Total de Nodos
  N = len(TARIFAS[0])

  #Inicialización de la tabla de precios
  PRECIOS = [ [9999]*N for i in [9999]*N]
  RUTA = [ [""]*N for i in [""]*N]

  for i in range(0,N-1):
    RUTA[i][i] = i             #Para ir de i a i se "pasa por i"
    PRECIOS[i][i] = 0          #Para ir de i a i se se paga 0
    for j in range(i+1, N):
      MIN = TARIFAS[i][j]
      RUTA[i][j] = i

      for k in range(i, j):
        if PRECIOS[i][k] + TARIFAS[k][j] < MIN:
            MIN = min(MIN, PRECIOS[i][k] + TARIFAS[k][j] )
            RUTA[i][j] = k          #Anota que para ir de i a j hay que pasar por k
        PRECIOS[i][j] = MIN

  return PRECIOS,RUTA
################################################################


PRECIOS,RUTA = Precios(TARIFAS)
#print(PRECIOS[0][6])

print("PRECIOS")
for i in range(len(TARIFAS)):
  print(PRECIOS[i])

print("\nRUTA")
for i in range(len(TARIFAS)):
  print(RUTA[i])

#Determinar la ruta con Recursividad
def calcular_ruta(RUTA, desde, hasta):
  if desde == hasta:
    #print("Ir a :" + str(desde))
    return ""
  else:
    return str(calcular_ruta( RUTA, desde, RUTA[desde][hasta])) +  \
                ',' + \
                str(RUTA[desde][hasta] \
              )

print("\nLa ruta es:")
calcular_ruta(RUTA, 0,6)
```

## Práctica Individual: Encontrar los dos puntos más cercanos

**Problema:** Dado un conjunto de puntos, encontrar la pareja con menor distancia.
Se pide:
1. Probar en 1D con Fuerza Bruta y calcular complejidad.
2. Mejorar en 1D con Divide y Vencerás.
3. Extender a 2D.
4. Extender a 3D.
5. Probar con conjuntos aleatorios y comparar tiempos.

### 1. Problema en 1D

* **Fuerza Bruta:** Compara cada número con todos los demás. Tiene dos bucles anidados, por lo que su complejidad es **O(n²)**.
* **Divide y Vencerás / Ordenación:** Si ordenamos la lista (complejidad O(n log n)), los dos números más cercanos tienen que estar juntos. Solo hace falta recorrer la lista una vez comparando vecinos (O(n)). Complejidad total: **O(n log n)**.

```python
import math

# 1D - Fuerza bruta O(n^2)
def puntos_cercanos_1d_fb(lista):
    min_dist = float('inf')
    mejor_par = None
    for i in range(len(lista)):
        for j in range(i + 1, len(lista)):
            d = abs(lista[i] - lista[j])
            if d < min_dist:
                min_dist = d
                mejor_par = (lista[i], lista[j])
    return min_dist, mejor_par

# 1D - Divide y Vencerás (ordenando la lista) O(n log n)
def puntos_cercanos_1d_dv(lista):
    ordenada = sorted(lista)
    min_dist = float('inf')
    mejor_par = None
    for i in range(len(ordenada) - 1):
        d = abs(ordenada[i] - ordenada[i + 1])
        if d < min_dist:
            min_dist = d
            mejor_par = (ordenada[i], ordenada[i + 1])
    return min_dist, mejor_par

# Prueba con la lista de la diapositiva
lista_1d = [3403, 4537, 9089, 9746, 7259, 1205, 4540, 8921]

print("1D Fuerza Bruta:     ", puntos_cercanos_1d_fb(lista_1d))
print("1D Divide y Vencerás:", puntos_cercanos_1d_dv(lista_1d))
```

### 2. Extensión a 2D

En 2D los puntos son `(x, y)` y la distancia es euclídea: $\sqrt{(x_1-x_2)^2 + (y_1-y_2)^2}$.

* **Fuerza Bruta:** Compara todas las parejas: complejidad **O(n²)**.
* **Divide y Vencerás:**
  1. Ordenamos los puntos por el eje X.
  2. Dividimos los puntos por la mitad y resolvemos cada mitad recursivamente.
  3. Tomamos la menor distancia `d = min(d_izq, d_der)`.
  4. Revisamos los puntos que están en la franja central a menos de `d` de la línea divisoria. En esa franja ordenamos por Y y solo hace falta comparar cada punto con los siguientes pocos vecinos (máximo 7 u 8).
  5. Complejidad: **O(n log n)**.

```python
# Distancia euclídea 2D
def dist_2d(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2)

# 2D - Fuerza bruta
def puntos_cercanos_2d_fb(puntos):
    min_dist = float('inf')
    mejor_par = None
    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            d = dist_2d(puntos[i], puntos[j])
            if d < min_dist:
                min_dist = d
                mejor_par = (puntos[i], puntos[j])
    return min_dist, mejor_par

# 2D - Divide y Vencerás
def puntos_cercanos_2d_dv(puntos):
    pts_x = sorted(puntos, key=lambda p: p[0])

    def resolver(pts):
        n = len(pts)
        if n <= 3:
            return puntos_cercanos_2d_fb(pts)

        mitad = n // 2
        medio_x = pts[mitad][0]

        d_izq, par_izq = resolver(pts[:mitad])
        d_der, par_der = resolver(pts[mitad:])

        if d_izq < d_der:
            d_min, mejor_par = d_izq, par_izq
        else:
            d_min, mejor_par = d_der, par_der

        # Franja central
        franja = [p for p in pts if abs(p[0] - medio_x) < d_min]
        franja.sort(key=lambda p: p[1])

        for i in range(len(franja)):
            for j in range(i + 1, min(i + 8, len(franja))):
                if (franja[j][1] - franja[i][1]) >= d_min:
                    break
                d = dist_2d(franja[i], franja[j])
                if d < d_min:
                    d_min = d
                    mejor_par = (franja[i], franja[j])

        return d_min, mejor_par

    return resolver(pts_x)

# Prueba 2D
puntos_2d = [(1122, 6175), (135, 4076), (7296, 2741), (1150, 6120), (5000, 5000)]
print("2D Fuerza Bruta:     ", puntos_cercanos_2d_fb(puntos_2d))
print("2D Divide y Vencerás:", puntos_cercanos_2d_dv(puntos_2d))
```

### 3. Extensión a 3D

En 3D los puntos son `(x, y, z)`. Seguimos la misma idea: dividir por el plano X y revisar la franja central en 3D.

```python
# Distancia euclídea 3D
def dist_3d(p1, p2):
    return math.sqrt((p1[0] - p2[0])**2 + (p1[1] - p2[1])**2 + (p1[2] - p2[2])**2)

# 3D - Fuerza bruta
def puntos_cercanos_3d_fb(puntos):
    min_dist = float('inf')
    mejor_par = None
    for i in range(len(puntos)):
        for j in range(i + 1, len(puntos)):
            d = dist_3d(puntos[i], puntos[j])
            if d < min_dist:
                min_dist = d
                mejor_par = (puntos[i], puntos[j])
    return min_dist, mejor_par

# 3D - Divide y Vencerás
def puntos_cercanos_3d_dv(puntos):
    pts_x = sorted(puntos, key=lambda p: p[0])

    def resolver(pts):
        n = len(pts)
        if n <= 3:
            return puntos_cercanos_3d_fb(pts)

        mitad = n // 2
        medio_x = pts[mitad][0]

        d_izq, par_izq = resolver(pts[:mitad])
        d_der, par_der = resolver(pts[mitad:])

        d_min, mejor_par = (d_izq, par_izq) if d_izq < d_der else (d_der, par_der)

        franja = [p for p in pts if abs(p[0] - medio_x) < d_min]
        franja.sort(key=lambda p: p[1])

        for i in range(len(franja)):
            for j in range(i + 1, min(i + 20, len(franja))):
                if (franja[j][1] - franja[i][1]) >= d_min:
                    break
                d = dist_3d(franja[i], franja[j])
                if d < d_min:
                    d_min = d
                    mejor_par = (franja[i], franja[j])

        return d_min, mejor_par

    return resolver(pts_x)

# Prueba 3D
puntos_3d = [(10, 20, 30), (100, 200, 300), (12, 22, 31), (45, 90, 12), (11, 21, 30.5)]
print("3D Fuerza Bruta:     ", puntos_cercanos_3d_fb(puntos_3d))
print("3D Divide y Vencerás:", puntos_cercanos_3d_dv(puntos_3d))
```

### 4. Pruebas con datos aleatorios y tiempos de ejecución

Generamos listas de puntos aleatorios para comparar el tiempo entre Fuerza Bruta y Divide y Vencerás.

```python
import random
import time

# Función para generar puntos aleatorios
def generar_datos_2d(n):
    return [(random.uniform(0, 10000), random.uniform(0, 10000)) for _ in range(n)]

print("Comparación de tiempos en 2D:")
for n in [200, 500, 1000, 2000]:
    datos = generar_datos_2d(n)

    t0 = time.time()
    res_fb = puntos_cercanos_2d_fb(datos)
    t_fb = time.time() - t0

    t0 = time.time()
    res_dv = puntos_cercanos_2d_dv(datos)
    t_dv = time.time() - t0

    print(f"N = {n:<5} | T. Fuerza Bruta: {t_fb:.5f}s | T. Divide y Vencerás: {t_dv:.5f}s")
```

### Conclusiones
* **Fuerza Bruta:** Es sencilla de implementar pero al ser **O(n²)** se vuelve muy lenta cuando $N$ crece (para miles de puntos ya tarda segundos y luego minutos).
* **Divide y Vencerás:** Al dividir el espacio y acotar las comparaciones en la franja central, baja la complejidad a **O(n log n)**. Como se ve en las pruebas, para $N=2000$ es mucho más rápido y permite trabajar con grandes cantidades de puntos sin problema.
